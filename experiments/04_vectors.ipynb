# 一组数怎样作为整体运算？

想象先向东 1、向北 2，再向东 3、向南 1。先预测合起来的位移，再逐个运行下面的单元格。

```python
import numpy as np

u = np.array([1, 2])
v = np.array([3, -1])
print('合位移 =', u + v)
```

第一坐标合并东西方向，第二坐标合并南北方向。现在预测：把 `u` 这段位移重复两次，和把两个坐标都乘以 2 一样吗？

```python
print('u + u =', u + u)
print('2 * u =', 2 * u)
print('两种计算一样吗？', np.array_equal(u + u, 2 * u))
```

## 把运算画出来

`u + v` 是合位移，`2u` 是把 `u` 延长两倍。箭头是直观场景；同样的坐标运算也可以表示信号、资源数量或方程组中的输入。

```python
import matplotlib.pyplot as plt

vectors = [(u, 'u'), (v, 'v'), (u + v, 'u + v'), (2 * u, '2u')]
fig, ax = plt.subplots(figsize=(6, 5))
for vector, label in vectors:
    ax.quiver(0, 0, *vector, angles='xy', scale_units='xy', scale=1, label=label)
ax.set(xlim=(-1, 5), ylim=(-2, 5), xlabel='first coordinate', ylabel='second coordinate', aspect='equal')
ax.axhline(0, color='0.75', linewidth=0.8)
ax.axvline(0, color='0.75', linewidth=0.8)
ax.grid(alpha=0.2)
ax.legend()
plt.show()
```
