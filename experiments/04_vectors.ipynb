# 先合并对应方向，再把记录看成整体

一段位移的东西、南北分量分别相加，但它们描述同一次变化。下面把整体相加和重复同一位移分别画出来。坐标顺序与正负约定必须一致；相同运算也能迁移到库存、采样值，不代表这些量都有相同物理含义。

从新内核按顺序运行；没有滑块时可直接修改参数。计算单元用于核对前面的推理，参考结果放在相应尝试之后。


```python
# 让图中的中文正常显示：先找本机已有的中文字体；找不到时（例如 Colab）下载思源黑体。
import pathlib
import urllib.request
import matplotlib.pyplot as plt
from matplotlib import font_manager

def use_chinese_font():
    candidates = ['Noto Sans CJK SC', 'Noto Sans SC', 'Source Han Sans SC', 'PingFang SC',
                  'Hiragino Sans GB', 'Microsoft YaHei', 'SimHei', 'WenQuanYi Zen Hei']
    installed = {font.name for font in font_manager.fontManager.ttflist}
    name = next((n for n in candidates if n in installed), None)
    if name is None:
        path = pathlib.Path.home() / '.cache' / 'linear-algebra-fonts' / 'NotoSansSC-Regular.otf'
        try:
            if not path.exists():
                path.parent.mkdir(parents=True, exist_ok=True)
                urllib.request.urlretrieve(
                    'https://github.com/notofonts/noto-cjk/raw/main/Sans/SubsetOTF/SC/'
                    'NotoSansSC-Regular.otf', path)
            font_manager.fontManager.addfont(str(path))
            name = font_manager.FontProperties(fname=str(path)).get_name()
        except Exception as error:
            print('没有找到中文字体，图中汉字可能显示为方框：', error)
            return
    plt.rcParams['font.family'] = [name, 'DejaVu Sans']
    plt.rcParams['axes.unicode_minus'] = False

use_chinese_font()
```

```python
import numpy as np

u = np.array([1, 2])
v = np.array([3, -1])
print('合位移 =', u + v)
```

第一坐标合并东西方向，第二坐标合并南北方向。现在预测：把 `u` 这段位移重复两次，和把两个坐标都乘以 2 一样吗？

```python
print('u + u =', u + u)
print('2 * u =', 2 * u)
print('两种计算一样吗？', np.array_equal(u + u, 2 * u))
```

## 把运算画出来

`u + v` 是合位移，`2u` 是把 `u` 延长两倍。箭头是直观场景；同样的坐标运算也可以表示信号、资源数量或方程组中的输入。

```python
import matplotlib.pyplot as plt

vectors = [
    (u, 'u', '#0072B2'),
    (v, 'v', '#D55E00'),
    (u + v, 'u + v', '#009E73'),
    (2 * u, '2u', '#CC79A7'),
]
fig, ax = plt.subplots(figsize=(6, 5))
for vector, label, color in vectors:
    ax.quiver(0, 0, *vector, angles='xy', scale_units='xy', scale=1,
              color=color, width=0.008, label=label)
ax.set(xlim=(-1, 5), ylim=(-2, 5), xlabel='第一个坐标', ylabel='第二个坐标', aspect='equal')
ax.axhline(0, color='0.75', linewidth=0.8)
ax.axvline(0, color='0.75', linewidth=0.8)
ax.grid(alpha=0.2)
ax.legend()
plt.show()
```
