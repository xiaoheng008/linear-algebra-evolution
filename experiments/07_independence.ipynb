# 多一列，范围扩大还是表示增多？

一列是另一列两倍时，多出的系数可能只制造不同配法。这里先找两份系数产生相同输出，再相减，读出一份非零调整为什么给零输出。相关描述的是这种调整存在，不是说原向量本身等于零。

```python
# 让图中的中文正常显示：先找本机已有的中文字体；找不到时（例如 Colab）下载思源黑体。
import pathlib
import urllib.request
import matplotlib.pyplot as plt
from matplotlib import font_manager

def use_chinese_font():
    candidates = ['Noto Sans CJK SC', 'Noto Sans SC', 'Source Han Sans SC', 'PingFang SC',
                  'Hiragino Sans GB', 'Microsoft YaHei', 'SimHei', 'WenQuanYi Zen Hei']
    installed = {font.name for font in font_manager.fontManager.ttflist}
    name = next((n for n in candidates if n in installed), None)
    if name is None:
        path = pathlib.Path.home() / '.cache' / 'linear-algebra-fonts' / 'NotoSansSC-Regular.otf'
        try:
            if not path.exists():
                path.parent.mkdir(parents=True, exist_ok=True)
                urllib.request.urlretrieve(
                    'https://github.com/notofonts/noto-cjk/raw/main/Sans/SubsetOTF/SC/'
                    'NotoSansSC-Regular.otf', path)
            font_manager.fontManager.addfont(str(path))
            name = font_manager.FontProperties(fname=str(path)).get_name()
        except Exception as error:
            print('没有找到中文字体，图中汉字可能显示为方框：', error)
            return
    plt.rcParams['font.family'] = [name, 'DejaVu Sans']
    plt.rcParams['axes.unicode_minus'] = False

use_chinese_font()
```

```python
import numpy as np

u = np.array([1, 2])
v = np.array([2, 4])

first = 1 * u + 0 * v
second = -1 * u + 1 * v
print('第一组系数 (1, 0) ->', first)
print('第二组系数 (-1, 1) ->', second)
print('输出相同吗？', np.array_equal(first, second))
print('v - 2u =', v - 2 * u)
```

## 系数变了，向量没变

沿着 `c1 + 2c2 = 1` 选系数，每一组都会组合出 `u`。左图画系数组合，右图画实际得到的向量。拖动 `t`，看系数点移动时输出是否跟着变。

```python
def show_representations(t=0.0):
    c1, c2 = 1 - 2 * t, t
    result = c1 * u + c2 * v
    line_t = np.linspace(-2, 2, 200)

    fig, (ax_coeff, ax_vector) = plt.subplots(1, 2, figsize=(9, 4))
    ax_coeff.plot(1 - 2 * line_t, line_t, color='#0072B2',
                  label='c1 + 2c2 = 1')
    ax_coeff.scatter([c1], [c2], color='#D55E00', s=70, zorder=3,
                     label='当前系数')
    ax_coeff.scatter([1, -1], [0, 1], color='#009E73', s=45, zorder=2,
                     label='(1, 0) 与 (-1, 1)')
    ax_coeff.set(xlabel='c1', ylabel='c2', title='不同的系数')
    ax_coeff.grid(alpha=0.2)
    ax_coeff.legend(fontsize=8)

    ax_vector.quiver(0, 0, *result, angles='xy', scale_units='xy', scale=1,
                    color='#0072B2', width=0.012)
    ax_vector.set(xlim=(-0.5, 1.5), ylim=(-0.5, 2.5), xlabel='输出第一项',
                  ylabel='输出第二项', title='相同的结果向量')
    ax_vector.set_aspect('equal', adjustable='box')
    ax_vector.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()
    print('系数 =', (c1, c2))
    print('c1*u + c2*v =', result)

import matplotlib.pyplot as plt
show_representations(t=0)  # 没有滑块时，修改 t 后重新运行这一行

try:
    import ipywidgets as widgets
    from IPython.display import display
except ImportError:
    print('未安装 ipywidgets，没有滑块；直接修改上面的函数参数也可以。')
else:
    display(widgets.interactive(
        show_representations,
        t=widgets.FloatSlider(value=0, min=-2, max=2, step=0.05, description='t')))
```
