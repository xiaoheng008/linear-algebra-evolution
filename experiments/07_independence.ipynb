# 两列向量，有没有重复信息？

令 `u = (1, 2)`，`v = (2, 4)`。先不看代码：如果输出是 `u`，你能找到两组不同的系数 `(c1, c2)`，使 `c1*u + c2*v = u` 吗？

```python
import numpy as np

u = np.array([1, 2])
v = np.array([2, 4])

first = 1 * u + 0 * v
second = -1 * u + 1 * v
print('first coefficients (1, 0) ->', first)
print('second coefficients (-1, 1) ->', second)
print('same output?', np.array_equal(first, second))
print('v - 2u =', v - 2 * u)
```

## 系数变了，向量没变

沿着 `c1 + 2c2 = 1` 选系数，每一组都会组合出 `u`。左图画系数组合，右图画实际得到的向量。拖动 `t`，看系数点移动时输出是否跟着变。

```python
def show_representations(t=0.0):
    c1, c2 = 1 - 2 * t, t
    result = c1 * u + c2 * v
    line_t = np.linspace(-2, 2, 200)

    fig, (ax_coeff, ax_vector) = plt.subplots(1, 2, figsize=(9, 4))
    ax_coeff.plot(1 - 2 * line_t, line_t, color='#0072B2',
                  label='c1 + 2c2 = 1')
    ax_coeff.scatter([c1], [c2], color='#D55E00', s=70, zorder=3,
                     label='current coefficients')
    ax_coeff.scatter([1, -1], [0, 1], color='#009E73', s=45, zorder=2,
                     label='(1, 0) and (-1, 1)')
    ax_coeff.set(xlabel='c1', ylabel='c2', title='Different coefficients')
    ax_coeff.grid(alpha=0.2)
    ax_coeff.legend(fontsize=8)

    ax_vector.quiver(0, 0, *result, angles='xy', scale_units='xy', scale=1,
                    color='#0072B2', width=0.012)
    ax_vector.set(xlim=(-0.5, 1.5), ylim=(-0.5, 2.5), xlabel='output x',
                  ylabel='output y', title='Same resulting vector')
    ax_vector.set_aspect('equal', adjustable='box')
    ax_vector.grid(alpha=0.2)
    plt.tight_layout()
    plt.show()
    print('coefficients =', (c1, c2))
    print('c1*u + c2*v =', result)

import matplotlib.pyplot as plt
show_representations(t=0)  # Without sliders, change t and rerun this line

try:
    import ipywidgets as widgets
    from IPython.display import display
except ImportError:
    print('For the slider, install ipywidgets; the editable call above works without it.')
else:
    display(widgets.interactive(
        show_representations,
        t=widgets.FloatSlider(value=0, min=-2, max=2, step=0.05, description='t')))
```
