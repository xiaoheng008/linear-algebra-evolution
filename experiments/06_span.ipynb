# 先判断固定关系，再尝试系数

固定 u=(1,2)、v=(2,4)。目标(3,6)容易达到，目标(3,5)呢？先写任意组合的两个坐标，判断它们是否始终满足一个关系。然后只把第二列末项改成3，试同一目标；一个目标成功还不是所有目标成功，最后再解任意(p,q)。

从新内核按顺序运行；没有滑块时可直接修改参数。计算单元用于核对前面的推理，参考结果放在相应尝试之后。


```python
import numpy as np
import matplotlib.pyplot as plt

u = np.array([1, 2])
b = np.array([3, 5])
coefficients = np.array([1, 1])
v_parallel = np.array([2, 4])
v_other_direction = np.array([2, 3])

print('u + v_parallel =', coefficients[0] * u + coefficients[1] * v_parallel)
print('Can this equal b?', np.array_equal(coefficients[0] * u + coefficients[1] * v_parallel, b))
print('u + v_other_direction =', u + v_other_direction)
assert np.array_equal(u + v_other_direction, b)
assert np.array_equal(3 * u, [3, 6])
```

改变系数没有打破旧输出的第二项等于第一项两倍的关系，换列却能到达(3,5)。下面图只抽样；全部目标是否可达仍要解方程。


```python
values = np.linspace(-3, 3, 19)
c1, c2 = np.meshgrid(values, values)
fig, axes = plt.subplots(1, 2, figsize=(10, 4.5))

for ax, v, title in zip(axes, [v_parallel, v_other_direction],
                        ['Parallel columns', 'Different directions']):
    out_x = c1 * u[0] + c2 * v[0]
    out_y = c1 * u[1] + c2 * v[1]
    ax.scatter(out_x, out_y, s=9, alpha=0.35, color='#0072B2', label='combinations')
    ax.scatter(*b, s=110, marker='*', color='#D55E00', edgecolor='black',
               linewidth=0.6, zorder=3, label='target b')
    ax.set(title=title, xlabel='output x', ylabel='output y',
           xlim=(-10, 10), ylim=(-10, 10), aspect='equal')
    ax.grid(alpha=0.2)
    ax.legend()

plt.tight_layout()
plt.show()
```

## 不用重试已经见过的答案

改目标为(1,0)：从 a+2b=1、2a+3b=0 求新的两系数，再将上面的目标变量 b 改成 [1,0] 运行图像。最后把右端换成任意(p,q)，解释为什么不再有旧直线限制。


```python
def inspect_combination(t=4.0, c1_value=1.0, c2_value=1.0):
    v = np.array([2.0, t])
    result = c1_value * u + c2_value * v
    error = result - b
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.quiver(0, 0, *u, angles='xy', scale_units='xy', scale=1,
              color='#0072B2', width=0.009, label='u')
    ax.quiver(0, 0, *v, angles='xy', scale_units='xy', scale=1,
              color='#009E73', width=0.009, label='v')
    ax.scatter(*result, color='#0072B2', s=65, label='chosen combination')
    ax.scatter(*b, color='#D55E00', marker='*', s=150, label='target b')
    ax.set(xlim=(-7, 8), ylim=(-8, 8), xlabel='output x', ylabel='output y',
           title='Change the coefficients and the second column')
    ax.set_aspect('equal', adjustable='box')
    ax.grid(alpha=0.2)
    ax.legend()
    plt.show()
    print('v =', v)
    print('c1*u + c2*v =', result)
    print('result - b =', error)

inspect_combination(4, 1, 1)  # No-widget fallback: edit these three values and rerun

try:
    import ipywidgets as widgets
    from IPython.display import display
except ImportError:
    print('For sliders, install ipywidgets; the editable function call above still works.')
else:
    display(widgets.interactive(
        inspect_combination,
        t=widgets.FloatSlider(value=4, min=2.5, max=5.5, step=0.05, description='v y'),
        c1_value=widgets.FloatSlider(value=1, min=-3, max=3, step=0.1, description='c1'),
        c2_value=widgets.FloatSlider(value=1, min=-3, max=3, step=0.1, description='c2')))
```
